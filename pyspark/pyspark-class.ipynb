# What is PySpark?

PySpark is a Python API for Apache Spark used to:

- Process large datasets
- Transform data
- Analyze data

In simple terms: **PySpark helps us work with big data efficiently**

---

# What is CRUD?

CRUD stands for:

| Operation | Meaning | PySpark Equivalent |
| --- | --- | --- |
| C | Create | Create DataFrame / Add rows |
| R | Read | Select / Filter |
| U | Update | withColumn |
| D | Delete | Filter (remove rows) |

---

# 1. CREATE → Add Data

### Purpose:

Create or add data into a DataFrame

```python
data = [
    (1,"Shreyas","UK","2024-01-10"),
    (2,"Priya","India","2025-04-25"),
    (3,"Pooja","UK","2024-03-27"),
    (4,"Rajesh","india","2026-01-30")
]

columns = ["customer_id","name","country","signup_date"]

df = spark.createDataFrame(data, columns)

df.display()
```

```python
#A PySpark DataFrame is immutable, so you don't directly insert a row into df. Instead, create another DataFrame for the new row and union it with the existing one.

new_data = [
    (5,"ravi gurjar","UK","2026-08-30")
]

new_df = spark.createDataFrame(new_data, columns)
```

```python
#combine two dataframe because you cannot add new data to existing dataframe
df = df.union(new_df)
df.display()
```

## Key Points:

- PySpark DataFrames are **immutable**
- You don’t modify → you create a new DataFrame

## Practice:

# 3. UPDATE → Modify Data

### Purpose:

Update existing data

```python
from pyspark.sql.functions import when

df = df.withColumn("country",
                   when(df.customer_id == 4,"Germany")
                   .otherwise(df.country)
                   )
df.display()
```

```python
#Adds a new column
#lit is litereal. i.e value
from pyspark.sql.functions import lit

df=df.withColumn("source", lit("manual"))

df.display()
```

### Important:

- No direct UPDATE like SQL
- You recreate the column

### Key Points:

- Always assign back to `df`
- Logic is applied row-wise

### Practice:

- Update country for one customer
- Update multiple customers

```python
df = df.withColumn("country",
 when(df.customer_id==1,"UK")
.when(df.customer_id==2,"USA")
.when(df.customer_id==3,"UK")
     .otherwise(df.country))

df.display()
```

# 12. UNION

Use `union()` or `unionByName()` to combine rows from two DataFrames. `unionByName()` matches columns by name and is generally safer when column order may differ.


```python
# Reload students DataFrame
df = spark.read.table("rivadataplatform.landing.students")

# Get schema from existing DataFrame
schema = df.schema

# Create new student data matching the schema
new_data = [("6", "Sarah Khan", "sarah@email.com", "London", "UK", "", "", "", "2024-06-01")]
new_df = spark.createDataFrame(new_data, schema)

df = df.unionByName(new_df)
df.display()
```
