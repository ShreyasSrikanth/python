## 1. Data types: different kinds of values

A data type tells Python what kind of value something is.

| Type | Example | Meaning |
| --- | --- | --- |
| `int` | `10` | Whole number |
| `float` | `10.5` | Number with decimals |
| `str` | `"Apple"` | Text |
| `bool` | `True` or `False` | Yes or no |

Use `type(value)` to ask Python what type a value has.

````python
item_name = "Notebook"
quantity = 3
price = 2.50
is_available = True

print(item_name, type(item_name))
print(quantity, type(quantity))
print(price, type(price))
print(is_available, type(is_available))
````

### Simple calculations and conversions

Numbers can be used in calculations. Text that looks like a number is still text until we convert it with `int()` or `float()`.

````python
text_quantity = "3"
quantity = int(text_quantity)
price = 2.50

total = quantity * price
print("Total:", total)
print("Quantity type:", type(quantity))
````

## 2. A list stores many values

A list is an ordered group of values. Lists are useful when we have many products or many sales.

We can add a value with `.append()` and read a value by its position. Python positions start at `0`.

````python
products = ["Notebook", "Pen", "Bag"]

print("First product:", products[0])
products.append("Pencil")
print("All products:", products)
````

### 3. A `for` loop visits every item

A `for` loop repeats the same instructions for each item in a list.

Read this as: **for each product in products, print the product**.

The variable `product` temporarily holds one item at a time.

````python
for product in products:
    print("Product:", product)
````

## 4. Three more useful collections

### Tuple: a fixed group of values

A tuple uses parentheses. It keeps values in order, but we normally do not change it.

```python
location = (10, 20)
```

### Set: unique values only

A set uses curly braces. Duplicate values are removed automatically.

```python
numbers = {1, 1, 2, 3}
```

### Dictionary: values with names

A dictionary stores a value beside a key. It is useful for describing one object, such as an order.

```python
order = {"customer": "Asha", "amount": 25}
```

We will look at each example in the next cell.

````python
# Tuple: ordered values that should stay together
location = (10, 20)
print("Tuple:", location)
print("First coordinate:", location[0])

# Set: duplicate values are kept only once
numbers = {1, 1, 2, 3}
print("Set:", numbers)
print("Number of unique values:", len(numbers))

# Dictionary: keys describe the values
order = {"customer": "Asha", "amount": 25}
print("Dictionary:", order)
print("Customer:", order["customer"])
order["amount"] = 30
print("Updated amount:", order["amount"])
````

## 4. Conditional statements

Conditions turn business rules into decisions. Comparisons include `>`, `<`, `==`, and `!=`; combine them with `and`, `or`, and `not`. Python executes the first matching branch.

````python
amount = 25

if amount > 20:
    print("This order is bigger than 20")
else:
    print("This order is 20 or smaller")
````

## 6. Use a loop to calculate a total

The loop can do more than print. We can keep a running total.

`total = total + price` means: take the old total, add the current price, and store the new total.

````python
prices = [2.50, 3.00, 1.50]
total = 0

for price in prices:
    total = total + price

print("Total:", total)
````

## 7. What is `assert`?

`assert` is a quick check for a result that should be true.

- If the condition is `True`, Python continues quietly.
- If the condition is `False`, Python stops and shows an error.

It is useful while learning and testing code. It is not a replacement for handling real user errors.

````python
prices = [2.50, 3.00, 1.50]
total = 0

for price in prices:
    total = total + price

assert total == 7.0
print("The check passed. The total is", total)
````

## 8. Functions: reusable instructions

A function is a named group of instructions. We define it once with `def` and call it whenever we need it.

`return` sends a result back to the code that called the function.

````python
def add(a, b):
    return a + b

result = add(3, 4)
print("The answer is", result)
assert result == 7
````

## 9. Small example: combine a list, a loop, and an `if`

This is a gentle recap. We will add only the prices of products that cost more than 2.

````python
prices = [1.50, 2.50, 3.00]
total = 0

for price in prices:
    if price > 2:
        total = total + price

print("Total of prices over 2:", total)
assert total == 5.5
````
