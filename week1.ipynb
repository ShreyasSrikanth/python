## Topic 5: Loops

### Question 1
Use a `for` loop to print each item in `[10, 20, 30]`.

```python
# Answer 1
# Write your answer here
li = [10,20,30]

for value in range(len(li)):
    print(li[value], end=" ")

print("\n")

for value in li:
    print(value, end=" ")

print("\n")

print(*li)
```

### Question 2
Use a loop to calculate the sum of `[2, 4, 6]`.

```python
# Answer 2
# Write your answer here

li = [2,4,6]

sum = 0

for value in li:
    sum = sum + value

print(f"The sum of the values within the list is {sum}")
```

### Question 3
Count how many values in `[5, 12, 8, 20]` are greater than 10.

```python
# Answer 3
# Write your answer here

li = [5,12,8,20]
count = 0

for value in li:
    if value > 10:
        count += 1

print(f"The number of values greater than 10 in the list is {count}")
```

### Question 3
Write a function `is_adult(age)` that returns `True` when age is at least 18.

```python
# Answer 3
# Write your answer here

def is_adult(age):
    if age >= 18:
        return True
    else:
        return False
    
shreyas = 29
rahul = 17

print(f"Shreyas is an adult : {is_adult(shreyas)}")
print(f"Rahul is an adult : {is_adult(rahul)}")
```
