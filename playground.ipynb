# Dictionary

```python


Dict = {
    1: "shreyas",
    2: "Ravi",
    3: "Ankur",
    4: "Jayraj",
    "string": "Random string value",
    5: {"Srikanth" : {
        28: "Accountant"
    }
        
}
    }

print(Dict)
```

```python
# This is how you access nested values
print("Nested dict 1 is",Dict[5]["Srikanth"])
print("Nested dict 1 is",Dict[5]["Srikanth"][28])
```
